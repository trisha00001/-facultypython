**📄 Основные принципы ФП.**

1. Функции являются полноценными объектами, их можно передавать как аргументы, возвращать друг из друга, присваивать переменным.

2. Неизменяемость. Все элементы кода считаются неизменяемыми: их значения после создания не меняются. Абсолютно все новые элементы создаются на основе существующих.

3. Отсутствие побочных эффектов. Функции не должны изменять внешнюю среду, например, глобальные переменные или иметь «скрытые» эффекты. Их выход определяется исключительно входными данными.

4. Рекурсия: способ разбиения проблемы на меньшие подзадачи, которые решаются с помощью одинаковых наборов функций.

5. Одни функции способны использовать другие в качестве своих аргументов, а затем возвращать их.

В целом, FP предлагает мощные концепции для надежного, легко читаемого, поддерживаемого кода.



###**📌Преимущества функционального программирования**

У ФП есть ряд преимуществ перед императивным стилем программирования. Перечислим основные.

1. Более простой код. ФП максимально использует чистые функции без побочных эффектов, что делает код более предсказуемым, читаемым, понятным. Отсутствие мутабельных компонентов упрощает отладку и рефакторинг.

2. Легкое тестирование. Чистые функции легко тестируются, так как их поведение не зависит от внешней среды.

3. Параллельное использование. FP часто используется для разработки параллельных и распределенных систем, так как неизменяемость исключает какие-либо проблемы с синхронным доступом для разных пользователей.

4. Эффективность. ФП позволяет оптимизировать код для повышения производительности, особенно в средах с ограниченными ресурсами.

5. Композиция. Функции легко компонуются друг с другом, что способствует повторному использованию кода.

6. Простота отладки кода, обусловленная чистотой, неизменяемостью, отсутствием побочных эффектов.

7. Улучшенная безопасность. Неизменяемость предотвращает случайные изменения, а также ошибки, связанные с одновременным доступом к данным.

8. Повышенная гибкость. ФП позволяет создавать более гибкий, адаптируемый код, который легко расширять и модифицировать.

**🔁5. Рекурсия и рекурсивные структуры данных**

**Рекурсия** — это мощный инструмент в FP, позволяющий разбить сложную задачу на более простые подзадачи, решаемые той же функцией. Рекурсивные структуры определяются с помощью самих себя. К ним относятся, например, списки и деревья.

Принцип рекурсии в FP состоит в том, что функция вызывает сама себя внутри своего определения, но с измененными входными данными.

```python
def factorial(n):

 if n == 0:

  return 1

 else:

  return n * factorial(n - 1)
```

♾️ **К рекурсивным структурам относятся:**

- списки — могут быть определены как пустой список или как элемент, добавленный к другому списку.
- деревья — могут быть определены как пустое дерево или как узел, содержащий данные и ссылки на другие деревья (потомки).

🎯**Преимущества рекурсии:**

- простота — рекурсивные решения часто более просты и понятны, чем итерационные.
- элегантность — рекурсивные решения красиво решают задачи, связанные с иерархическими данными.
- совместимость с ФП — рекурсия естественным образом вписывается в парадигму функционального программирования.


🔸 Рекурсия требует внимательного подхода, чтобы избежать бесконечных циклов. Необходимо убедиться, что рекурсивная функция имеет базовый случай — условие, которое останавливает рекурсию.

В Python функция len() сама по себе не принимает параметра key, однако для некоторых функций, таких как max() и min(), параметр key действительно используется для применения пользовательской функции к каждому элементу перед сравнением.


Давай разберём, как работает key в контексте функций max() и min(), а затем как его можно применить с len() в этих функциях.

###**🧠 Использование key с len() в Python**
###📖 Что такое key?

Параметр **key** позволяет передать функцию, которая будет применяться к каждому элементу коллекции перед тем, как выполнить операцию. Это полезно, если нужно сравнивать не сами элементы, а их какие-то характеристики (например, длину строки или значение в словаре).



###**🔧 Пример использования key с len():**
###1. max() и min() с key на основе длины строк

Если у нас есть список строк, и мы хотим найти строку с наибольшей или наименьшей длиной, мы можем использовать key=len:

```python
#key=len заставляет функции max() и min() сравнивать строки не по их значению, а по длине.
words = ['apple', 'banana', 'kiwi', 'cherry']

# Найдём строку с наибольшей длиной
longest_word = max(words, key=len)
print(longest_word)  # 'banana'

# Найдём строку с наименьшей длиной
shortest_word = min(words, key=len)
print(shortest_word)  # 'kiwi'
```

 ### Пример с кортежами

Предположим, у нас есть список кортежей, где каждый кортеж состоит из имени и возраста. Мы можем использовать len() как часть key, чтобы найти кортеж с наибольшим/наименьшим именем:

```python
people = [('Alice', 25), ('Bob', 22), ('Charlie', 30)]

# Найдём человека с самым длинным именем
longest_name_person = max(people, key=lambda person: len(person[0]))
print(longest_name_person)  # ('Charlie', 30)

# Найдём человека с самым коротким именем
shortest_name_person = min(people, key=lambda person: len(person[0]))
print(shortest_name_person)  # ('Bob', 22)
# Здесь используется lambda person: len(person[0]), чтобы вычислить длину имени для каждого человека.
```

###**🔑 Как работает key?**

Функция, переданная в параметр key, применяется к каждому элементу и возвращает значение, которое будет использовано для сравнения.

Например, в случае с len() будет возвращена длина строки или любого другого итерируемого объекта, и уже по этим длинам будет происходить сравнение.

###**🚀 Применение key для других целей**

Сортировка по длине строк:

```python
words = ['apple', 'banana', 'kiwi', 'cherry']
sorted_words = sorted(words, key=len)
print(sorted_words)  # ['kiwi', 'apple', 'banana', 'cherry']
```

Поиск максимальной длины (если тебе нужно просто значение длины):

```python
words = ['apple', 'banana', 'kiwi', 'cherry']
max_len = max(words, key=len)
print(len(max_len))  # 6
```

Когда элементы сложно сравнивать напрямую, и нужно подготовить их перед сравнением

###**✅ all() и any() в Python**

Обе функции работают с итерируемыми объектами (списками, кортежами, генераторами и т.д.) и проверяют логическое значение элементов — как булевые фильтры.



all() и any() - это встроенные функции Python, которые позволяют сделать удобные проверки boolean для итерируемого объекта. all() возвращает True, если все элементы в итерируемом объекте равны True. Это эквивалентно следующему коду:

```python
def all(iterable):
    for element in iterable:
        if not element:
            return False
    return True
```

any() возвращает True, если любой из элементов итерируемого объекта равен True. Это эквивалентно следующему коду:

```python
def any(iterable):
    for element in iterable:
        if element:
            return True
    return False
```

###**🟩 all() — «Все True?»**
Возвращает True, если все элементы в итерируемом объекте — истинны (True).

```python
print(all([True, True, True]))  # 👉 True
print(all([True, False, True]))  # 👉 False
```

💬 Метафора: "И"

🔌 Все лампочки должны быть включены, чтобы комната загорелась.

```python
numbers = [2, 4, 6]
print(all(n % 2 == 0 for n in numbers))  # 👉 True (все чётные)
```

###**🟨 any() — «Хоть один True?»**

Возвращает True, если хотя бы один элемент — истинен (True).


```python
print(any([False, False, True]))  # 👉 True
print(any([False, False, False]))  # 👉 False
```

💬 Метафора: "ИЛИ"

🔌 Хоть одна лампочка загорелась — и уже светло.

```python
numbers = [1, 3, 4]
print(any(n % 2 == 0 for n in numbers))  # 👉 True (нашлась одна чётная)
```

###**🔢 enumerate() — нумерует элементы**
### 📖 Что делает?

Функция enumerate() добавляет индексы к элементам итерируемого объекта.
Она возвращает итерируемый объект из пар (индекс, значение).

 Enumerate предоставляет счётчик, который увеличивается во время итераций по объекту. Результат возвращается в виде кортежа (count,element). По сути, эта функция эквивалентна следующему коду:

```python
def enumerate(sequence, start=0):
    n = start
    for elem in sequence:
        yield n, elem
        n += 1
```

🔹 iterable — любая последовательность (список, строка и т.д.)

🔹 start — с какого номера начать счёт (по умолчанию 0)

###**🔗 zip() — связывает несколько последовательностей**

###📖 Что делает?

Функция zip() объединяет элементы из двух и более последовательностей в кортежи — поэлементно.

###**🔧 Синтаксис:**

```python
zip(seq1, seq2, ...)
```

zip() создаёт итератор, который агрегирует элементы из итерируемых объектов.

Эта функция возвращает итератор из кортежей, где i-ый кортеж содержит i-ые элементы из каждой последовательности или итерируемого объекта, переданных в качестве параметров. Итератор останавливается, когда мы доходим до конца самого короткого из итерируемых объектов. Если итерируемый объект один, то функция возвращает кортежи из одного элемента. Если вызвать функцию без параметров, то она вернёт пустой итератор.

zip() эквивалентен следующему коду:

```python
def zip(*iterables):
    # zip('ABCD', 'xy') --> Ax By
    sentinel = object()
    iterators = [iter(it) for it in iterables]
    while iterators:
        result = []
        for it in iterators:
            elem = next(it, sentinel)
            if elem is sentinel:
                return
            result.append(elem)
        yield tuple(result)
```

zip() допустимо использовать для объектов неодинаковой длины только в случае, если Вам не важны "в хвосте" более длинных итерируемых объектов, до которых не дойдёт дело.

**📌 Используется для создания пар, словарей или объединённых данных.**

```python
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]

for name, age in zip(names, ages):
    print(name, age)
```

### **🗺️ map() — Переобразователь массива**
**📖 Что это?**

map() — это встроенная функция в Python, которая позволяет применить функцию к каждому элементу итерируемого объекта (например, списка) и получить новый результат.

 | 💬 Просто: "Сделай что-то с каждым элементом и верни результат".



**map()** - это встроенная функция Python, которая принимает на вход два или более параметра: функцию, и один или несколько итерируемых объектов, в следующем виде:

**🔧 Синтаксис:**

*map(function, iterable, ...)*

🔹function (функция) — что делать с каждым элементом

🔹 iterable (итерируемый_объект) — список, кортеж, строка и т.д.

🔹 возвращает итератор (в Python 3), который можно превратить в список с помощью list()



**map()** возвращает iterator - специальный объект, который постепенно возвращает результат по мере необходимости. Мы узнаем больше об итераторах и генераторах в будущих лекциях. **map() с несколькими итераторами**

map() может работать с более чем одним итератором. Итераторы должны быть одной и той же длины - в случае, если это не так, то map() остановится, как только самый короткий из итераторов дойдёт до конца.

Например, если наша функция складывает два значения x и y, то мы можем передать список значений x и другой список значений y в функцию map(). Функция (или лямбда) будет запущена для нулевых элементов из каждого списка, затем для первых элементов из каждого списка, и так далее до n-ых элементов.

Давайте посмотрим это в действии сначала для двух, а затем для трёх списков:


```python
a = [1,2,3,4]
b = [5,6,7,8]
c = [9,10,11,12]

list(map(lambda x,y:x+y,a,b))
```

```python
# Now all three lists
list(map(lambda x,y,z:x+y+z,a,b,c))
```


В примере выше видно, что параметр x получает свои значения из списка a, параметр y получает значения из b и z из списка c. Попробуйте поэкспериментировать сами с каким-нибудь своим примером, чтобы убедиться, что Вы понимаете работу map с более чем одним итерируемым объектом.

```python
l1 = [0,1,2,3,4]
l2 = [10,9,8,7,6]

def multi(d1, d2):
  return d1 * d2

result = map(multi, l1, l2)
print( list(result) )
```

###**🔍 filter() — отбор по условию**
###📖 Что делает?

filter() — это встроенная функция Python, которая отбирает только те элементы из последовательности, для которых функция возвращает True.

💬 Проще говоря:
_“Оставь только те, кто проходит проверку”._

###**🔧 Синтаксис:**



```python
filter(функция, итерируемый_объект)
```

🔹 функция — возвращает True или False

🔹 итерируемый_объект — список, кортеж и т.д.

🔹 возвращает итератор → оберни в list() или tuple(), если нужен результат

Функция filter(function, list) предоставляет удобный способ отфильтровать элементы итерируемого объекта и оставить только те из них, для которых указанная в параметре функция возвращает True.

В качестве первого параметра filter(function,list) принимает на вход функцию, которая возвращает значение Boolean (True или False). Эта функция применяется к каждому элементу итерируемого объекта. Если функция возвращает True, то элемент добавляется в результат.

Как и в случае с map(), filter() возвращает итератор - то есть, filter возвращает один результат за другим, по мере необходимости. Итераторы и генераторы будут изучены в одной из ближайших лекций. Здесь же, поскольку объёмы данных в наших примерах малы, мы можем преобразовать filter() в список и получить результаты сразу же.

```python

# Сначала создадим функцию
def even_check(num):
    if num%2 ==0:
        return True
```

```python
lst =range(20)

list(filter(even_check,lst))
```

```python
list(filter(lambda x: x%2==0,lst))
```

```python
numbers = [1, 2, 3, 4, 5, 6]

even = filter(lambda x: x % 2 == 0, numbers)

print(list(even))  # [2, 4, 6]
```

➡️ lambda x: x % 2 == 0 — оставляет только чётные числа

```python
words = ['apple', '', 'banana', '', 'kiwi']

non_empty = filter(None, words)  # None — проверяет на "истинность"

print(list(non_empty))  # ['apple', 'banana', 'kiwi']
```

###**🔽 reduce() — свёртка/накопление значений**

###📖 Что это?
**reduce()** — функция, которая последовательно применяет заданную функцию к элементам последовательности, сводя всё к одному результату.

**💬 Простыми словами:**

Берёт два элемента → применяет функцию → берёт результат и следующий элемент → снова применяет функцию → … пока не останется одно значение.

###**🧠 Где живёт reduce?**
Она находится в модуле functools, поэтому сначала нужно:

```python
from functools import reduce
```

###**🔧 Синтаксис:**

```python
reduce(функция, итерируемый_объект, [начальное_значение])
```

🔹 функция — принимает два аргумента

🔹 итерируемый_объект — список, кортеж и т.д.

🔹 начальное_значение (опционально) — значение, с которого начинается свёртка

###**🔍 Пример 1: сумма всех чисел**

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

result = reduce(lambda x, y: x + y, numbers)

print(result)  # 15
```

➡️ Как работает:

```python
1 + 2 → 3
3 + 3 → 6
6 + 4 → 10
10 + 5 → 15
```

###**🔍 Пример 2: умножение**

```python
from functools import reduce

result = reduce(lambda x, y: x * y, [1, 2, 3, 4])
print(result)  # 24
```

```python
result = reduce(lambda x, y: x +'!'+ y, ['1', '2', '3', '4'])
print(result)  # 24
```

Представь снежный ком, который катится по горе и собирает всё по пути ⛄

Или как сложение элементов по цепочке:

###**🆚 Когда лучше использовать reduce:**

🔹 Если нужно получить одно значение из множества (например: сумма, произведение, объединение)

🔹 Если lambda + reduce читается нормально (иначе — лучше for или sum())

🔹 Может быть менее читаемо, чем обычный for

🔹 Для многих задач лучше использовать готовые функции, как sum() или max()
